```python
import torch
```

```python
rand_sample = torch.rand(10000)
print("Rand sample:", rand_sample)
randn_sample = torch.randn(10000)
print("Randn sample:", randn_sample)
zeros_sample = torch.zeros(10)
print("Zeros sample:", zeros_sample)
ones_sample = torch.ones(10)
print("Ones sample:", ones_sample)
arange_sample = torch.arange(0, 10)
print("Arange sample:", arange_sample)
linspace_sample = torch.linspace(0, 10, steps=5)
print("Linspace sample:", linspace_sample)
eye_sample = torch.eye(5)
print("Indentity sample:", eye_sample)
empty_sample = torch.empty(10)
print("Empty sample:", empty_sample)
```

```python
# Basic Arithmetic Operations
x = torch.tensor([5.0, 2.0, 4.0])
y = torch.tensor([2.0, 1.0, 3.0])

print(x+y)
print(x*y)
print(x@y) # vector dot product
# aggregations
print(x.sum())
print(x.mean())
print(x.max())
```

```python
# Activation Functions
import torch.nn.functional as F
x = torch.tensor([-2.0, -1.0, 0.0, 1.0, 2.0])
# relu - sets negative values to 0
print(F.relu(x))
# sigmoid - normalises values between 0 and 1
print(F.sigmoid(x))
# tanh - normalises values between -1 and 1
print(F.tanh(x))
```

```python
x = torch.tensor(3.0, requires_grad=True) # enable gradient tracking
f = x**2
# .backward() -> calculates gradient using backpropagation
f.backward()
# check gradient(derivative) at point x
print(x.grad)
```
